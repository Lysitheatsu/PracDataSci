```python
import pandas as pd

abalone_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/abalone/abalone.data"
col_names = ['Sex', 'Length', 'Diameter', 'Height', 'Whole weight','Shucked weight', 'Viscera weight', 'Shell weight', 'Rings']
abalone_df = pd.read_csv(abalone_url, header=None, names=col_names)
```

```python
abalone_df.groupby('Sex')['Rings'].max()
```

```python
infant_df = abalone_df[abalone_df['Sex'] == 'I']
heaviest_infant = infant_df.loc[infant_df['Whole weight'].idxmax()]
heaviest_infant['Rings']
```
