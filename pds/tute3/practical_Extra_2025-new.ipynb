## Create a DataFrame by customized data

```python
# Create two-dimensional, size-mutable, potentially heterogeneous tabular data by yourself.
# Please understand how pd.DataFrame works via:
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html

# {'key1': [], 'key2': [], ...} is a dictionary in Python

import pandas as pd
df1 = pd.DataFrame({'name': ['Jason', 'Lucy', 'Helen'], 'height': [180, 170, 165], 'weight': [75, 65, 72]}) 
df1
```

```python
df2 = pd.DataFrame({'name': ['Jason', 'Lucy', 'Helen'], 'mathematics': [80, 90, 65], 'science': [75, 65, 72]})
df2
```

```python
df3 = pd.DataFrame({'name': ['Jason', 'Lucy', 'Helen'], 'company': ['RMIT', 'RMIT', 'RMIT'], 'salary': [100000, '120,000', '110,000']})
df3
```

```python
df4 = pd.DataFrame({'name': ['Jason', 'Lucy', 'Helen'], '# of attendance': [22, 22, 15]})
df4
```

## Merge different DataFrames by 'name'

```python
# Merge DataFrame or named Series objects with a database-style join.
# Please understand how merge() function works via:
# https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html#pandas.DataFrame.merge

merged_df = df1.merge(df2, on='name', how='inner')
merged_df = merged_df.merge(df3, on='name', how='inner')
merged_df = merged_df.merge(df4, on='name', how='inner')
merged_df
```

```python
# Use Method Chain

merged_df = df1.merge(df2, on='name', how='inner').merge(df3, on='name', how='inner').merge(df4, on='name', how='inner')
merged_df
```

```python
# Use For loop

all_df = [df2, df3, df4]
merged_df = df1
for df in all_df:
    merged_df = merged_df.merge(df, on='name', how='inner')
merged_df
```

## Add new observation to merged_df

```python
merged_df
```

```python
# A part of Helen's attendance is recorded into another dataset, which needs to be added to merged_df
new_df = pd.DataFrame([{'name': 'Helen', 'height': 165, 'weight': 72, 'mathematics': 65, 
                              'science': 72, 'company': 'RMIT', 
                              'salary': '110,000', '# of attendance': 6}])
```

```python
merged_df = pd.concat([merged_df, new_df], ignore_index=True)
```

## The rows with index 2 and 3 are recorded separately for the same staff, which needs to merge to one row.

```python
# groupby(): Group DataFrame using a mapper or by a Series of columns.
# Please refer to: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html

# transform(): Call function producing a like-indexed Series on each group and return a Series 
# having the same indexes as the original object filled with the transformed values.
# Please refer to: 
# https://pandas.pydata.org/docs/reference/api/pandas.core.groupby.SeriesGroupBy.transform.html

merged_df['# of attendance'] = merged_df.groupby(['name', 'height', 'weight', 'mathematics', 'science', 
                                                  'company', 'salary'])['# of attendance'].transform('sum') # use the return values of sum() function as the transformed values
merged_df['# of attendance']
```

## Remove duplicates

```python
merged_df
```

```python
# Return DataFrame with duplicate rows removed.
# Please refer to: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop_duplicates.html

merged_df = merged_df.drop_duplicates(subset=['name', 'height', 'weight', 'mathematics', 'science', 
                                              'company', 'salary']).reset_index(drop=True)
merged_df
```

## Transfer 'salary' to numbers

```python
# Use astype to transform data type
# astype(): Cast a pandas object to a specified dtype dtype.
# Please refer to: https://pandas.pydata.org/docs/reference/api/pandas.Series.astype.html

# str.replace(): Replace each occurrence of pattern/regex in the Series/Index.
# Please refer to: https://pandas.pydata.org/docs/reference/api/pandas.Series.str.replace.html

merged_df['salary'] = merged_df['salary'].astype('str').str.replace(',', '').astype('int')
merged_df
```

```python
# Or use pd.to_numeric
# Convert argument to a numeric type.
# Please refer to: https://pandas.pydata.org/docs/reference/api/pandas.to_numeric.html

merged_df['salary'] = pd.to_numeric(merged_df['salary'].astype('str').str.replace(',', ''))
merged_df
```
