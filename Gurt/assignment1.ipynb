# Task 1: Data Preparation


```python
# Start from here to write your code for Task 1. Please also feel free to add more cells in this ipynb file.
import pandas as pd
import numpy as np
import re

# Load the original CSV into a DataFrame.
df = pd.read_csv("victorian_road_crash_data.csv")

# Verify the loaded data matches the source file by checking shape and column names.
print(f"Loaded data shape: {df.shape[0]} rows x {df.shape[1]} columns")
print(f"Columns: {list(df.columns)}")
```

```python
# Convert ACCIDENT_DATE and ACCIDENT_TIME to proper datetime formats for easier analysis.
df['ACCIDENT_DATE'] = pd.to_datetime(df['ACCIDENT_DATE'], format='%d/%m/%Y', errors='coerce')
df['ACCIDENT_TIME'] = pd.to_datetime(df['ACCIDENT_TIME'], format='%H:%M:%S', errors='coerce').dt.time
```

```python
# Remove duplicate rows to ensure data integrity.
df = df.drop_duplicates()
print(f"Rows after removing duplicates: {len(df)}")

# Drop the SRNS column as it is missing a large portion of values (~70%).
df = df.drop(columns=['SRNS'])

# Drop rows with missing values in critical columns (small portion of the data).
df = df.dropna(subset=['LATITUDE', 'LONGITUDE', 'NO_OF_VEHICLES'])

# Fill missing categorical values with 'Unknown' to retain the rows.
categorical_cols = ['ROAD_NAME', 'ROAD_TYPE', 'LGA_NAME', 'DTP_REGION',
                    'DEG_URBAN_NAME', 'RMA', 'DIVIDED', 'STAT_DIV_NAME']
df[categorical_cols] = df[categorical_cols].fillna('Unknown')
```

```python
# Drop the redundant VICGRID coordinate columns since LATITUDE and LONGITUDE are available.
df = df.drop(columns=['VICGRID_X', 'VICGRID_Y'], errors='ignore')
```

```python
# Clean the SPEED_ZONE column and extract numeric speed values.
def clean_speed(val):
    if pd.isna(val):
        return np.nan
    val = str(val).lower()
    if 'not know' in val or 'other' in val:
        return np.nan
    # Use regex to extract the leading numeric value.
    nums = re.findall(r'\d+', val)
    if nums:
        speed = int(nums[0])
        # Remove impossible speed values (0 km/h or over 110 km/h).
        if speed > 110 or speed == 0:
            return np.nan
        return speed
    return np.nan

# Apply cleaning and store as a nullable integer column.
df['SPEED_ZONE_NUM'] = df['SPEED_ZONE'].apply(clean_speed).astype('Int64')
```

```python
# Convert vehicle count columns from float to integer for correct representation.
cols_to_fix = [
    'NO_OF_VEHICLES', 'HEAVYVEHICLE', 'PASSENGERVEHICLE',
    'MOTORCYCLE', 'PT_VEHICLE'
]
for col in cols_to_fix:
    df[col] = df[col].astype(int)

# Fill missing ROAD_ROUTE_1 values with 0 before converting to integer.
df['ROAD_ROUTE_1'] = df['ROAD_ROUTE_1'].fillna(0).astype(int)
```

```python
# Replace underscores with spaces in DEG_URBAN_NAME for better readability.
df['DEG_URBAN_NAME'] = df['DEG_URBAN_NAME'].str.replace('_', ' ', regex=False)
```

```python
# Display the final cleaned dataset shape and preview the first few rows for verification.
print(f"Data successfully cleaned. Final shape: {df.shape}")
print(f"\nMissing values remaining:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
display(df.head())
```

```python
df.to_csv('cleaned.csv', index=False)
```
